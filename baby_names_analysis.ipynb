````
"""
===============================================================================
Baby Names Popularity Analysis (2007 → 2020)
Assignment: Find which names from the 2007 Top 10 fell out of the Top 100 in 2020
Author   : Pramod Mayigoowda
Date     : November 2025
Files downloaded from SSA using this link https://www.ssa.gov/OACT/babynames/index.html and converted into a comma separated text file
    - yob2007.txt   → SSA baby names for 2007 (with header: name,count,gender)
    - yob2020.txt   → SSA baby names for 2020 (with header: name,count,gender)

How to run:
    1. Place this script and the two .txt files in the same folder
    2. Run:   python baby_names_analysis.py
    3. The program will print the Top 10 from 2007 and their exact 2020 ranks,
       then clearly list the names that fell outside the Top 100.

Result: From the data I downloaded, I could not find any names which fell out of top 100 in 2020. So, I printed top 10 names in 2007 for male
and female and also their 2020 rank
===============================================================================
"""

import pandas as pd

# Load your files (header present, columns: name, count, gender)
df2007 = pd.read_csv("/Users/pramod/Desktop/yob2007.txt")
df2020 = pd.read_csv("/Users/pramod/Desktop/yob2020.txt")

# Separate by gender and rank by count (descending)
m2007 = df2007[df2007['gender'] == 'M'].sort_values('count', ascending=False).reset_index(drop=True)
f2007 = df2007[df2007['gender'] == 'F'].sort_values('count', ascending=False).reset_index(drop=True)
m2020 = df2020[df2020['gender'] == 'M'].sort_values('count', ascending=False).reset_index(drop=True)
f2020 = df2020[df2020['gender'] == 'F'].sort_values('count', ascending=False).reset_index(drop=True)

# Assign ranks
m2007['rank_2007'] = m2007.index + 1
f2007['rank_2007'] = f2007.index + 1
m2020['rank_2020'] = m2020.index + 1
f2020['rank_2020'] = f2020.index + 1

# Top 10 in 2007
top10_male_2007 = m2007.head(10)[['name', 'rank_2007', 'count']].copy()
top10_female_2007 = f2007.head(10)[['name', 'rank_2007', 'count']].copy()

# Lookup dicts for 2020 ranks
rank2020_male = dict(zip(m2020['name'], m2020['rank_2020']))
rank2020_female = dict(zip(f2020['name'], f2020['rank_2020']))

# Add 2020 ranks (9999 if missing)
top10_male_2007['rank_2020'] = top10_male_2007['name'].map(rank2020_male).fillna(9999).astype(int)
top10_female_2007['rank_2020'] = top10_female_2007['name'].map(rank2020_female).fillna(9999).astype(int)

# Identify fallen (rank >100)
fallen_males = top10_male_2007[top10_male_2007['rank_2020'] > 100]
fallen_females = top10_female_2007[top10_female_2007['rank_2020'] > 100]

# Print results
print("Top 10 Male 2007 with 2020 ranks:")
print(top10_male_2007[['name', 'rank_2007', 'rank_2020']])
print("\nTop 10 Female 2007 with 2020 ranks:")
print(top10_female_2007[['name', 'rank_2007', 'rank_2020']])
print("\nFallen Males:", fallen_males['name'].tolist())
print("\nFallen Females:", fallen_females['name'].tolist())
````
